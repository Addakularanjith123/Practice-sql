```python
#prime or not using while loop
n=int(input("enter n value"))
c=0
i=1
while(i<=n):
  #c=0
  if n%i==0:
    c=c+1
  i=i+1
if c==2:
  print("the number is prime",+n)
else:
  print("the number is not a prime",+n)
```

```python
#prime from m to n using while loop
m=int(input("enter starting value "))
n=int(input("enter ending value"))
print("the prime numbers from",m,"to",n,"are:")
while(m<=n):
  c=0
  j=1
  while(j<=m):
    if m%j==0:
      c=c+1
    j=j+1
  if c==2:
    print(j)
  m=m+1
```

```python
#factorial of the given number using for loop
n=int(input("enter n value"))
fact=1
count=0
for i in range (1,n+1):
  fact=fact*i
  count=fact+count
print(fact)
```

```python
#factorial from m to n using for loop
m=int(input("enter m value:"))
n=int(input("enter n value"))
print("the numbers from",m,"to",n,"are:")
#fact=1
for i in range(m,n+1):
  fact=1
  for j in range(1,i+1):
    fact=fact*j
  print(fact, end="\n")
```

```python
#factorial from m to n using while loop
m=int(input("enter m value"))
n=int(input("enter n value"))
print("the factorial of the numbers from",m,"to",n,"are:")
#i=1
while(m<=n):
  fact=1
  j=1
  while(j<=m):
    fact=fact*j
    j=j+1
  print(fact,end=" ")
  m=m+1
```
