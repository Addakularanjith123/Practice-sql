###**<font color="blue"><b><h2 align="first">prime or not </font> (Check given number using while)**

```python
#prime or not using while loop
n=int(input("enter n value"))
i=1
c=0
while(n>=i):
    # c=0 # if the n=7 then the 'c' value will be(0(initial value),1,0,0,0,0,1) it won't be 2 for our prime requirement i.e it will always gives the not a prime values only
    if n%i==0:
        c=c+1
    i=i+1
# i=i+1 ##if we write this condition it will be outside of the while loop it gives continueous loop like object c will be keep on incrementing c=1,2,3,4,5,6,7.....(infinite loop)
if c==2:
    print("the number",n,"is prime")
    # print("prime")
else:
    # print("the number",n,"is not a prime")
    print("not prime")
```

###**<font color="blue"><b><h2 align="first">prime or not </font> (Check given number using for loop)**

```python
n=int(input("enter n value"))
c=0
for i in range(1,n+1):
    if n%i==0:
        c=c+1
if c==2:
    print("prime")
else:
    print("not prime")
```

###**<font color="blue"><b><h2 align="center">prime or not upto the range</font> (Check given range(n) of number)**

```python
#prime or not from the range using for loop
n=int(input("enter upto how many values"))
print("the prime numbers upto", n,"are:")
for i in range(1,n+1):
    c=0
    for j in range(1,i+1):
        if i%j==0:
            c=c+1
    if c==2:
        # print(i,sep='/t')
        print(i,end=' ')
```
