```python
missing = dim_cust['email'].isna() #storing missing values in 'missing' object like in data record 2 it true

random_numbers = pd.Series(
    np.random.randint(0, 100, size=missing.sum()), #np.random.randint(start, stop, size) here there are 382 missing records
    index=dim_cust.index[missing] #select only missing indexes of the data frame eg.0,1,2,3,4,5,6,7...only 1,3,7 indexes are missing the email(i.e has Nan values) then it will only select those indexes Index([1, 3, 7]).
).astype(str).str.zfill(2) #converting into string datpe because to append and str.zfill(2) it make every number only 2 string digit

dim_cust.loc[missing, 'email'] = (    #loc is used to select the row and columns  and Select the email column only where missing == True.
    dim_cust.loc[missing, 'full_name'] #take the full_name values from rows where missing is True, and put them into the email column of those same rows
        .str.strip() #removing the spaces like first and last spaces
        .str.lower() #converting them into lower case
        .str.replace(r'\s+', '.', regex=True) #r'\s+' (Find one or more whitespace characters.)
                                    #replacing space with '.(dot)' and + used for one or more white spaces like('Ranjith  Addakula) there are two space then it will give(ranjith.addakula)
    + random_numbers #adding the above object
    + '@outlook.com' #adding the fixed domain
)
```

```python
import numpy as np

missing = df['email'].isna()

domains = np.random.choice(
    ['hotmail.com', 'outlook.com', 'gmail.com', 'yahoo.com'],
    size=missing.sum()
)

random_numbers = np.random.randint(0, 100, size=missing.sum())

df.loc[missing, 'email'] = (
    df.loc[missing, 'full_name']
    .str.lower()
    .str.replace(' ', '.', regex=False)
    + np.char.zfill(random_numbers.astype(str), 2)
    + '@'
    + domains
)
```

```python
df_cust=pd.read_csv('/content/drive/MyDrive/Kaggle datasets/Lotus Retail Dataset - Power BI (Multi tables)/dim_customers.csv')
```

```python
import numpy as np

missing = df_cust['email'].isna()

domains = np.random.choice(
    ['hotmail.com', 'outlook.com', 'gmail.com', 'yahoo.com'],
    size=missing.sum()
)

phone_digits = df_cust.loc[missing, 'phone'].astype(str).apply(
    lambda x: ''.join(np.random.choice(list(set(x)), 2, replace=False))
)

df_cust.loc[missing, 'email'] = (
    df_cust.loc[missing, 'full_name']
    .str.lower()
    .str.replace(' ', '.', regex=False)
    + phone_digits
    + '@'
    + domains
)
```
